```python
import asyncio
from pyrpl import Pyrpl
from pyrpl.async_utils import sleep

p = Pyrpl(config='', hostname='_FAKE_')
```

```python
p.rp.scope.setup(duration=0.0001, trace_average=1)
p.rp.scope.continuous()
await asyncio.sleep(0.05)
curve = p.rp.scope.single()
```

```python
assert curve.shape == (2, 2**14)
```

```python
from qtpy import QtCore, QtTest
```

```python
p.lockbox.classname = "Interferometer"
```

```python
scope_widget = p.rp.scope._create_widget()
btn_single = scope_widget.button_single
btn_calibrate = p.lockbox.inputs['port1']._create_widget().button_calibrate
```

```python
import time
assert btn_single.isEnabled()
btn_single.click()
deadline = time.monotonic() + 3.0
while p.rp.scope.running_state != 'stopped' and time.monotonic() < deadline:
    time.sleep(0.01)
assert p.rp.scope.running_state == 'stopped'
from qtpy import QtWidgets
QtWidgets.QApplication.instance().processEvents()
scope_widget.button_continuous.click()
deadline = time.monotonic() + 3.0
while p.rp.scope.running_state != 'running_continuous' and time.monotonic() < deadline:
    time.sleep(0.01)
assert p.rp.scope.running_state == 'running_continuous'
p.rp.scope.stop()
assert btn_calibrate.isEnabled()
btn_calibrate.click()
deadline = time.monotonic() + 5.0
while not btn_calibrate.isEnabled() and time.monotonic() < deadline:
    QtWidgets.QApplication.instance().processEvents()
    time.sleep(0.01)
assert btn_calibrate.isEnabled()
```
